## CoinMarketCap API Quick Start

```python
from requests import Request, Session
from requests.exceptions import ConnectionError, Timeout, TooManyRedirects
import json

url = 'https://pro-api.coinmarketcap.com/v1/cryptocurrency/listings/latest'
parameters = {
  'start':'1',
  'limit':'5000',
  'convert':'USD'
}
headers = {
  'Accepts': 'application/json',
  'X-CMC_PRO_API_KEY': '28dff26c-34c1-4b94-82f9-bb62a21e486f',
}

session = Session()
session.headers.update(headers)

try:
  response = session.get(url, params=parameters)
  data = json.loads(response.text)
  #print(data)
except (ConnectionError, Timeout, TooManyRedirects) as e:
  print(e)
```

## Display Settings

```python
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)

# lambda is a shorthand function. x is its parameter. The lambda function returns the result of: '%.5f' % x
# In %.5f, % is a filler variable. The % x part means that % represents x.
# %.5f means the number that % represents will be formatted as a float with 5 decimal places.
# set_option('display.float_format', function) takes a function as its second parameter. We use a lambda function for simplicity.
pd.set_option('display.float_format', lambda x: '%.5f' % x)
```

## Function to Run API and Record Data in CSV

```python
def run_api():
    url = 'https://pro-api.coinmarketcap.com/v1/cryptocurrency/listings/latest'
    parameters = {'start':'1', 'limit':'10', 'convert':'USD'}
    headers = {'Accepts': 'application/json', 'X-CMC_PRO_API_KEY': '28dff26c-34c1-4b94-82f9-bb62a21e486f'}
    
    session = Session()
    session.headers.update(headers)
    
    try:
        response = session.get(url, params=parameters)
        data = json.loads(response.text)
        #print(data)
    except (ConnectionError, Timeout, TooManyRedirects) as e:
        print(e)

    df = pd.json_normalize(data['data'])
    df['timestamp'] = pd.to_datetime('now')

    # Check if file exists
    file_name = r'C:\Users\cenny\Programming\Data Analyist Bootcamp\Python\Crypto Data.csv'
    if not os.path.isfile(file_name):
        df.to_csv(file_name, header='column_names')
    else:
        df.to_csv(file_name, mode='a', header=False)
```

## Record Data for 5 min:
### The idea is to run this indefinitely in the background to collect data.

```python
i = 0
while i < 5:
    run_api()
    print('Ran API successfully')
    sleep(60) # 1 minute
    i += 1

exit()
```

```python
df2 = pd.read_csv(r"C:\Users\cenny\Programming\Data Analyist Bootcamp\Python\Crypto Web Scraping\Crypto Data.csv")
```

## Average Percent Changes of Crypto Currencies

```python
df3 = df2.groupby('name', sort=False)[['quote.USD.percent_change_1h', 'quote.USD.percent_change_24h', 'quote.USD.percent_change_7d', 'quote.USD.percent_change_30d', 'quote.USD.percent_change_60d', 'quote.USD.percent_change_90d']].mean(numeric_only=True)
df3
```

## Display Data as Stack

```python
df3.stack()
```

```python
type(df3.stack())
```

## Store Stack as Data Frame

```python
df4 = df3.stack().to_frame(name='values')
df4
```

```python
type(df4)
```

## Change Data Frame Labels

```python
df5 = df4.reset_index()
df5 = df5.rename(columns={'level_1': 'percent_change'})
df5
```

```python
df5['percent_change'] = df5['percent_change'].replace(['quote.USD.percent_change_1h', 'quote.USD.percent_change_24h', 'quote.USD.percent_change_7d', 'quote.USD.percent_change_30d', 'quote.USD.percent_change_60d', 'quote.USD.percent_change_90d'], ['1h', '24h', '7d', '30d', '60d', '90d'])
df5
```

## Plot Data

```python
sns.catplot(x='percent_change', y='values', hue='name', data=df5, kind='point')
```
